1. Single-Qubit Gate Order
Construct and compare the following two circuits:
Circuit A
H → X → Z
Circuit B
Z → X → H
For both circuits:
● Draw the circuit.
● Derive the state after every gate.
● Obtain the final statevector using Qiskit.
● Run the circuit for 1024 shots.
● Display the measurement counts and histogram.
● Compare the final states

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram


# -----------------------------
# Circuit A: H → X
# -----------------------------
qc_a = QuantumCircuit(1,1)

# -----------------------------
# Display Circuit A
# -----------------------------
print("Circuit A: H → X")


qc_a.h(0)
state1 = Statevector.from_instruction(qc_a)

qc_a.x(0)
state2 = Statevector.from_instruction(qc_a)

qc_a.z(0)
# Calculate final state
state_a = Statevector.from_instruction(qc_a)

display(qc_a.draw("mpl"))

print("State after first gate:")
print(state1)
print("State after second gate:")
print(state2)
print("Final State:")
print(state_a)

sim = AerSimulator()
qc_a.measure(0,0)
result = sim.run(qc_a, shots=1024).result()

counts = result.get_counts()

print("Measurement Counts for Circuit A:")
print(counts)

# Display histogram
display(plot_histogram(counts))

# -----------------------------
# Circuit B: X → H
# -----------------------------
qc_b = QuantumCircuit(1,1)


# -----------------------------
# Display Circuit B
# -----------------------------
print("\nCircuit B: X → H")


qc_b.z(0)

state1 = Statevector.from_instruction(qc_b)


qc_b.x(0)

state2 = Statevector.from_instruction(qc_b)


qc_b.h(0)

# Calculate final state
state_b = Statevector.from_instruction(qc_b)

display(qc_b.draw("mpl"))
print("State after first gate:")
print(state1)
print("State after second gate:")
print(state2)
print("Final State:")
print(state_b)

sim = AerSimulator()
qc_b.measure(0,0)
result = sim.run(qc_b, shots=1024).result()

counts = result.get_counts()

print("Measurement Counts for Circuit B:")
print(counts)

# Display histogram
display(plot_histogram(counts))
```

Construct circuits for all four Bell states:

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator
from IPython.display import display

print("Bell Gate 1")
sim = AerSimulator()

qc_a = QuantumCircuit(2,2)

qc_a.h(0)
qc_a.cx(0,1)

display(qc_a.draw("mpl"))

state = Statevector.from_instruction(qc_a)
print("Statevector:")
print(state)

qc_a.measure([0,1],[0,1])

result = sim.run(qc_a, shots=1024).result()
counts = result.get_counts()

print("\nCounts:")
print(counts)

display(plot_histogram(counts))



print("Bell Gate 2")
qc_b = QuantumCircuit(2,2)

qc_b.x(0)
qc_b.h(0)
qc_b.cx(0,1)

display(qc_b.draw("mpl"))

state = Statevector.from_instruction(qc_b)
print("Statevector:")
print(state)

qc_b.measure([0,1],[0,1])

result = sim.run(qc_b, shots=1024).result()
counts = result.get_counts()

print("\nCounts:")
print(counts)

display(plot_histogram(counts))


print("Bell State 3")
qc_c = QuantumCircuit(2,2)

qc_c.h(0)
qc_c.x(1)
qc_c.cx(0,1)


display(qc_c.draw("mpl"))

state = Statevector.from_instruction(qc_c)
print("Statevector:")
print(state)

qc_c.measure([0,1],[0,1])

result = sim.run(qc_c, shots=1024).result()
counts = result.get_counts()

print("\nCounts:")
print(counts)

display(plot_histogram(counts))

print("Bell State 4")


qc_d = QuantumCircuit(2,2)


qc_d.h(0)
qc_d.x(1)
qc_d.cx(0,1)
qc_d.z(1)

display(qc_d.draw("mpl"))

state = Statevector.from_instruction(qc_d)
print("Statevector:")
print(state)

qc_d.measure([0,1],[0,1])

result = sim.run(qc_d, shots=1024).result()
counts = result.get_counts()

print("\nCounts:")
print(counts)

display(plot_histogram(counts))
```

6. Entanglement Detective — Final Challenge
You are given the following three circuits.

Circuit 1

q₀: ── H ──●────
│
q₁: ───────X────

Circuit 2

q₀: ── H ──
q₁: ── H ──

Circuit 3

q₀: ── X ── H ──●────
│
q₁: ─────── H ──X────

For each circuit:
● Implement it in Qiskit.
● Obtain the statevector.
● Run 1024 shots.
● Display the measurement counts.
● Plot the histogram.
● Determine whether the final state is entangled or separable.
● Explain how you reached your conclusion.
Important: Do not determine entanglement solely from whether the measurement results
appear correlated. Use the quantum state and circuit structure in your explanation.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

# ------------------------------------
# Create a 2-Qubit Quantum Circuit
# ------------------------------------

qc_a = QuantumCircuit(2, 2)
print("Circuit A:")
# Put BOTH qubits into superposition
qc_a.h(0)
qc_a.cx(0,1)

# Calculate final state
state_a = Statevector.from_instruction(qc_a)
print("Final State:")
print(state_a)

# Measure both qubits
qc_a.measure([0,1], [0,1])

# Display the circuit
print("Quantum Circuit:\n")
display(qc_a.draw("mpl"))




# ------------------------------------
# Run the simulation
# ------------------------------------

sim = AerSimulator()

result = sim.run(qc_a, shots=1024).result()

counts = result.get_counts()

print("Measurement Counts:")
print(counts)

# Display histogram
display(plot_histogram(counts))


print("Circuit B")
qc_b = QuantumCircuit(2, 2)

# Put BOTH qubits into superposition
qc_b.h(0)
qc_b.h(1)

# Calculate final state
state_b = Statevector.from_instruction(qc_b)
print("Final State:")
print(state_b)

# Measure both qubits
qc_b.measure([0,1], [0,1])

# Display the circuit
print("Quantum Circuit:\n")
display(qc_b.draw("mpl"))




# ------------------------------------
# Run the simulation
# ------------------------------------

sim = AerSimulator()

result = sim.run(qc_b, shots=1024).result()

counts = result.get_counts()

print("Measurement Counts:")
print(counts)

# Display histogram
display(plot_histogram(counts))

print("Circuit C")
qc_c = QuantumCircuit(2, 2)

# Put BOTH qubits into superposition
qc_c.x(0)
qc_c.h(0)
qc_c.h(1)
qc_c.cx(0,1)

# Calculate final state
state_c = Statevector.from_instruction(qc_c)
print("Final State:")
print(state_c)

# Measure both qubits
qc_c.measure([0,1], [0,1])

# Display the circuit
print("Quantum Circuit:\n")
display(qc_c.draw("mpl"))




# ------------------------------------
# Run the simulation
# ------------------------------------

sim = AerSimulator()

result = sim.run(qc_c, shots=1024).result()

counts = result.get_counts()

print("Measurement Counts:")
print(counts)

# Display histogram
display(plot_histogram(counts))
```
